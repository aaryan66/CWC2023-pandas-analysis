
# ICC Cricket World Cup 2023 - EDA project

An Exploratory Data Analytics Project on Cricket World Cup 2023 where we look at the stats across games and get insights


```python
import pandas as pd

#csv -> pd dataframe
cwc = pd.read_csv("CWC2023.csv")

#before cleaning the data we hvae 48 entries including the null values if there are any
print(cwc.info())


#to see if there is any null values

print(cwc.isnull().sum()) 
# there are no null values implying that all the 48 rows contains matches with winners and result and no game had ended on N/R
# further checking was done via points table and the results were consistent with the data in the csv file

```

```python
#first up would be toss result analysis and how it affects the match result


#to see if teams prefer to bat or bowl first after winning the toss and whether the decision is right or wrong in terms of winning the match

batFirst = len(cwc[cwc['Toss Decision'] == "Bat"])/len(cwc) * 100
bowlFirst = len(cwc[cwc['Toss Decision'] == "Field"])/len(cwc) * 100

print(f"Percent of teams that chose to \n bat first : {round(batFirst,3)}% \n bowl first : {round(bowlFirst,3)}%")

# to see if the toss decision is right or wrong in terms of winning the match
```

```python
# Toss Winner Decision vs Match Result


#Since it was found that 54% of teams chose to field first and 46% chose to bat first, we can see if the decision was right or wrong in terms of winning the match

matchResult = len(cwc[cwc['Toss Winner'] == cwc['Wining Team']]) / len(cwc) * 100
batting_Selected = cwc[(cwc['Toss Decision'] == 'Bat') & (cwc['Toss Winner'] == cwc['Wining Team'])]
bowling_Selected = cwc[(cwc['Toss Decision'] == 'Field') & (cwc['Toss Winner'] == cwc['Wining Team'])]

BattingWinner = len(cwc[cwc['Team A'] == cwc['Wining Team']]) / len(cwc) * 100
FieldingWinner = len(cwc[cwc["Team B"] == cwc["Wining Team"]]) / len(cwc) * 100
BattingTossWinner = len(cwc[(cwc['Toss Winner'] == cwc['Wining Team']) & (cwc['Toss Decision'] == "Bat")])/len(cwc[cwc['Toss Decision'] == "Bat"]) * 100
FieldingTossWinner = len(cwc[(cwc['Toss Winner'] == cwc['Wining Team']) & (cwc['Toss Decision'] == 'Field')])/len(cwc[cwc['Toss Decision'] == "Field"]) * 100

print(f"Win % of team who won the toss: {round(matchResult,3)}%")
print(f"Win % of team Batting first: {round(BattingWinner, 3)}%")
print(f"Win % of team bowling first: {round(FieldingWinner, 3)}%")
print(f"Win % of team batting first after deciding to do so {round(BattingTossWinner, 3)}%")
print(f"Win % of team bowling first after deciding to do so {round(FieldingTossWinner, 3)}%")


print(f"Total Wins of Team who chose to Bat: {len(batting_Selected)} \n Total Wins of teams who chose to field: {len(bowling_Selected)}")
```
